Text Loaders

```python
from langchain_community.document_loaders import TextLoader
from langchain_openai import ChatOpenAI
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import PromptTemplate
from dotenv import load_dotenv
load_dotenv()
```

```python
loader = TextLoader("cricket.txt",encoding = "utf8")
documents = loader.load()
```

```python
model = ChatOpenAI()
prompt = PromptTemplate(
    template="Write summary of the following text:\n\n{text}\n\nSummary:",
    input_variables=["texts"],
)

parser = StrOutputParser()

chain = prompt | model | parser

print(chain.invoke({"text":documents[0].page_content}))
```

```python
print(type(documents[0]))
# print(documents[0].page_content)
print(f"Numer of documents loaded: {len(documents)}")
```

```python
print(f"Number of documents loaded form web : {len(documents_from_web)}")
print('='*60)
content = documents_from_web[0].page_content
```

```python
print(content)
```

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from dotenv import load_dotenv
load_dotenv()
```

```python
model = ChatOpenAI()
prompt = PromptTemplate(
    template="Answer the following question based on the provided text:\n\nText: {text}\n\nQuestion: {question}\n\nAnswer:",
    input_variables=["question","text"],
)
```

```python
parser = StrOutputParser()

chain = prompt| model | parser

print(
    chain.invoke(
        {'question':"what is the main topic of the trxt ?",'text':content}
    )
)
```

Directory Loaders 

```python
from langchain_community.document_loaders import DirectoryLoader, PyPDFLoader
import time

# normal / eager loading of documents from a directory
loader = DirectoryLoader(
    path="books",
    glob="**/*.pdf",
    loader_cls=PyPDFLoader
)

start = time.time()
documents = loader.load()
for doc in documents:
    print(f"=" * 60) 
    print(doc.page_content)
    print(f"=" * 60) 
end = time.time()
print(f"Time taken to load documents: {end - start} seconds")

# lazy loading of documents from a directory
lazy_loader = DirectoryLoader(
    path="books",
    glob="**/*.pdf",
    loader_cls=PyPDFLoader
)

documents_lazy = lazy_loader.lazy_load()
start = time.time()
for doc in documents_lazy:
    print(f"=" * 60)
    print(doc.page_content)
    print(f"=" * 60)
end = time.time()
print(f"Time taken to lazy load documents: {end - start} seconds")
```
